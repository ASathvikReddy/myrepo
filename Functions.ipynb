```python
# Passing Argumens to a function
# Positional Arguments
def pets(animal,name):
  print(f"\nI have a {animal} in  my house")
  print(f"And the name of my {animal} is {name.title()}")

animal="dog"
name="dev"
pets(animal,name)
pets("cat","gunter")
pets("cow","lakshmi")
```

```python
# Keyword Arguments
def pets(animal,name):
  print(f"\nI have a {animal} in  my house")
  print(f"And the name of my {animal} is {name.title()}")

pets(animal="dog",name="dev")
pets(name="gunter",animal="cat")
```

```python
# Returning a Dictionary
def build_person(first,last,age=None):
  person={"first_name":first,"last_name":last}
  if age: person['age']=age
  return person

user1=build_person("sathvik","Alla")
user2=build_person("sathvik","Alla",age=25)
print(user1,user2,sep="\n")
```

```python
def formatting_name(first, last):
    full_name = f"{first} {last}"
    return full_name.title()

while True:
    first = input("\nEnter your first name (or q to quit): ")
    if first == 'q':
        break
    last = input("Enter your last name (or q to quit): ")
    if last == 'q':
        break
    print(f"Hello, {formatting_name(first, last)} !!")
```

```python
def print_models(unprinted_designs,completed_models):
  while unprinted_designs:
    current_design=unprinted_designs.pop()
    print(f"Printing model: {current_design}")
    completed_models.append(current_design)

def show_completed_models(completed_models):
  print("\nThese are the completed Design Models : ")
  for i in completed_models:
    print(i)
unprinted_designs=["phone case","car charger","mouse pad","hat"]
completed_models=[]
#print(unprinted_designs)

#print_models(unprinted_designs,completed_models)
#show_completed_models(completed_models)
#print(f"The unprinted_designs left are : {unprinted_designs}")

# If we need the list of unprinted designs we need to pass in function as:
print_models(unprinted_designs[:],completed_models)
show_completed_models(completed_models)

print(f"The unprinted_designs : {unprinted_designs}")

# passing list[:] tells to access the copy of the original
```

```python
# Using *ARGS

def making_pizza(crust,*toppings):
  print(f"\n We are making a {crust} pizza with the following toppings :")
  for topping in toppings:
    print(f"- {topping}")

making_pizza("thick","pepperoni")
making_pizza("Thin","mushrooms","green peppers","extra cheese")
```

```python
#Using **KWARGS

def student_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key.title()} : {value}")
student_details(name="sathvik",age=25,location="pathapadu")
```

```python
def pizza_order(*args,**kwargs):
  print("Pizza Toppings are : ")
  for topping in args:
    print(f"- {topping}")

  print("\nOrder Details Are : ")
  for key,value in kwargs.items():
    print(f"{key.title()} : {value}")

pizza_order(
    "Cheese", "Mushroom", "Paneer",
    name="Satvik",
    size="Large",
    address="Vijayawada"
)
```
