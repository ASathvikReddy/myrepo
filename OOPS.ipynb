# **Encapsulation**

```python
class BankAccount:
  def __init__(self,balance):
    self.__balance = balance

  def deposit(self,amount):
    if amount > 0:
      self.__balance += amount

  def withdraw(self,amount):
    if amount > 0 and amount <= self.__balance:
      self.__balance -= amount

  def get_balance(self):
    return self.__balance

my_account = BankAccount(10000)

my_account.deposit(5000)
my_account.withdraw(2000)

print(f"My Account Balance is : {my_account.get_balance()}")
```

```python
# Encapsulation and Method Types

class BankAccount:
  def __init__(self,owner,balance=0):
    self.owner = owner
    self._balance = balance  # balance is now "proctected" by convention

  @property
  def show_balance(self):
    print(f"your balance is : {self._balance}")       # Read-Only Access

  def deposit(self,amount):
    if amount > 0:
      self._balance += amount

  def withdraw(self,amount):
    if amount > self._balance:
      print("Insufficient Amount")
    else:
      self._balance -= amount

  @classmethod
  def from_string(cls,text):        # Alternative Constructure can access or to modity the class data
    owner ,bal = text.split(',')
    return cls(owner,float(bal))

  @staticmethod
  def is_valid_amount(x):
    return x > 0

user = BankAccount('sathvik',10000)
user.deposit(5000)
user.withdraw(2000)
user.show_balance
```

# **Inheritance**

```python
class Employee:

  def __init__(self,name):
    self.name = name

  def login(self):
    print(f"{self.name} logged in")

class Developer(Employee):

  def works(self):
    print(f"{self.name} is working on a project")

class Tester(Employee):

  def tests(self):
    print(f"{self.name} is testing the code")

dev = Developer('sathvik')
test = Tester('reddy')

dev.login()
dev.works()

test.login()
test.tests()
```

**Inheritance & Polymorphism**

```python
class Employee:
    def __init__(self, name, salary):
        self.name, self.salary = name, salary

    def annual_pay(self):
        return self.salary * 12

class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)    # reuse parent init
        self.bonus = bonus

    def annual_pay(self):                 # override
        return super().annual_pay() + self.bonus

for e in [Employee("Sathvik", 50000), Manager("Reddy", 90_000, 200_000)]:
    print(e.name, e.annual_pay())         # polymorphism
```

# **Abstraction**

```python
# Abstraction means hiding implementation details and exposing only essential functionality.
# Python provides the abc module to implement abstract classes.

from abc import ABC,abstractmethod
class Vehicle(ABC):

  @abstractmethod
  def start(self):
    pass

class Car(Vehicle):
  def start(self):
    print("Car has started")

class Bike(Vehicle):
  def start(self):
    print("Bike has started")

car = Car()
car.start()

bike = Bike()
bike.start()
```
