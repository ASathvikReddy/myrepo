**Gussing Game**

```python

import random

while True:
    number = random.randint(1, 10)

    guess = int(input("I'm thinking of a number! Try to guess the number I'm thinking of: "))

    while guess != number:

        if guess < number:
            guess = int(input("Too low! Guess again: "))

        elif guess > number:
            guess = int(input("Too high! Guess again: "))

    play_again = input("That's it! Would you like to play again? (yes/no) ").lower()

    if play_again != "yes":
        print("Thanks for playing!")
        break
```

```python
users={
    "username":"sathvik",
    "first name":"Alla",
    "last name":"Reddy"
}

# looping through key-value pairs
for key,value in users.items():
  print(f"\n key :{key}")
  print(f" Value :{value}")
print()

# looping through all keys in dictionery
for name in users.keys():
  print(name.title(),end=" ")
print()

# looping through all values in dictionery
for name in users.values():
  print(name.title(),end="     ")
```

```python
hero_0={"name":"captian america","points":10}
hero_1={"name":"iron man","points":15}
hero_2={"name":"thor","points":20}
avengers=[hero_0,hero_1,hero_2]

for hero in avengers:
  print(hero)

for _ in range(10):
  new_hero={"name":"black panther","points":25}
  avengers.append(new_hero)
for i in avengers:
  print(i,end="\n")

for hero in avengers[:8]:
  if hero["name"]=="black panther":
    hero["name"]="hulk"
    hero["points"]=25
for i in avengers:
  print(i,end="\n")
```
